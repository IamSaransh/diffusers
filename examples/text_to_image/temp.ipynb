```python
import os

def remove_whitespace_from_filenames(directory):
    """Removes white spaces from all file names in the given directory."""
    if not os.path.exists(directory):
        print("Directory does not exist.")
        return

    for filename in os.listdir(directory):
        old_path = os.path.join(directory, filename)
        if os.path.isfile(old_path):  # Ensure it's a file, not a directory
            new_filename = filename.replace(" ", "_")  # Replace spaces with underscores
            new_path = os.path.join(directory, new_filename)

            if old_path != new_path:  # Avoid renaming if no change
                os.rename(old_path, new_path)
                print(f"Renamed: {filename} -> {new_filename}")

# Example Usage
remove_whitespace_from_filenames(dir)
```
